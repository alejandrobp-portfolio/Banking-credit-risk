```python
df_clean = df.copy()

# Eliminar duplicados exactos
df_clean = df_clean.drop_duplicates()

# Eliminar edades claramente imposibles
df_clean = df_clean[df_clean["person_age"] <= 100]
```

```python
# Eliminar antigüedades laborales imposibles
df_clean = df_clean[
    (df_clean["person_emp_length"].isna()) |
    (df_clean["person_emp_length"] <= df_clean["person_age"])
]
```

```python
# Eliminar antigüedades laborales imposibles
df_clean = df_clean[
    (df_clean["person_emp_length"].isna()) |
    (df_clean["person_emp_length"] <= df_clean["person_age"])
]
```
